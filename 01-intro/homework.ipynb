```python
# Median before filling
median_before = df["horsepower"].median()
print("Median before:", median_before)

# Most frequent value
most_frequent = df["horsepower"].mode()[0]
print("Most frequent:", most_frequent)

# Fill missing values and save the change
df["horsepower"] = df["horsepower"].fillna(most_frequent)
print("Missing after filling:", df["horsepower"].isna().sum())

# Median after filling
median_after = df["horsepower"].median()
print("Median after:", median_after)
```

```python
#Sum of weights
asian_cars = df[df["origin"] == "Asia"]
asian_car = asian_cars[["vehicle_weight", "model_year"]].head(7)
X = asian_car.values

XTX = X.T @ X
XTX_inverse = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inverse @ X.T @ y

print("w:", w.round(4))
print("Sum:", round(w.sum(), 4))
```
